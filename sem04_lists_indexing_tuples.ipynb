## **Списки (`list`)**

Типы данных в питоне делятся на два типа:
*   неизменяемые – `int`, `float`, `str`, `bool` (с ними мы уже знакомы) + `tuple` (с ним знакомимся сегодня)
*   изменяемые – `list` (с ним знакомимся сегодня) и `dict`, `set` (с ними познакомимся в течение модуля)



**Список – это упорядоченная изменяемая коллекция объектов любых типов.** При помощи него мы храним **набор** данных (числа, строки и т.д.) в **определенном** порядке. Данные мы можем менять – добавлять или удалять.



Зачем нужен список? Хранить **набор** данных (числа, строки и т.д.) в **определенном** порядке – с этим одна или две переменные, зарезервированные под одно значение, не справятся.

### Как создать список?

Встроенная функция `list()` либо создаст пустой список, либо сделает его из любого итерируемого объекта – в нем данные можно перебрать циклом.

````python
# просто пустой список
lst = list()
print(lst)
````

Квадратные скобки `[]`

````python
# просто пустой список
lst = []
print(lst)
````

````python
# готовые списки
guests = ['Вася', 'Татуся', 'Пётр Яковлевич', 'мама']  # список строк
digits = [1, 2, 3, 4, 5, 6, 7, 8, 9]  # список чисел
mixed_list = ['Вася', 45, True]  # в одном списке могут храниться элементы разных типов
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]  # список списков
super_mixed_list = [1, 2.0, '3, 4, 5', [6, 7, [8]]]  # список списков с разными типами данных
````

````python
print(*guests)
````

### Индексация и `len()`

У каждого элемента в списке есть свой **индекс** – номер, под которым элемент стоит в наборе.

**Индексация в питоне с нуля** – «первый» элемент имеет индекс 0.

К каждому элементу можно обратиться по его номеру при помощи квадратных скобок.

````python
list_of_wishes = ['iPhone', 'MacBook', 'Свитшот']

print(list_of_wishes[1])
````

Узнать число элементов в списке можно при помощи функции `len()`.

````python
print(len(list_of_wishes))
````

Какой будет индекс у последнего элемента, если индексация с нуля?

````python
print(list_of_wishes[len(list_of_wishes) - 1])
````

Покороче это можно записать при помощи **отрицательных** чисел – у последнего элемента будет индекс `-1`.

````python
print(list_of_wishes[-1])
````

Значение элемента списка по индексу можно поменять.

````python
list_of_wishes[-1] = 'Отпуск на море'
print(list_of_wishes)
````

### Как добавить элементы в список?

Обычно мы используем списки для **хранения** какой-то информации. Сначала создаем пустой список, а потом добавляем по ходу задачи туда значения.

Это можно сделать при помощи метода `.append()`.

````python
list_of_wishes = []  # cписок моих желаний
list_of_wishes.append('Прокатиться на американских горках')  # после минутных раздумий
print(list_of_wishes)
````

### Новый способ вводить данные: метод `split()`

Вообще-то это метод не списков, а строк – но они очень тесно связаны. Метод `split()` по заданному разделителю (в данном случае – пробел) сделает из строки список.

````python
digits = "1 2 3 4 5".split()
print(digits)
````

````python
digits = "1,2,3,4,5".split(",")
print(digits)
````

Если в практике данные вводятся в строку – мы используем именно `split()`.

````python
digits = input().split()
print(digits)
````

Если вводится несколько значений, и мы хотим каждое из них записать в **отдельную переменную**, то мы тоже так можем сделать при помощи `split()`.

````python
first, second = "1 2".split()
````

### Методы списков

Полная документация – [здесь](https://docs.python.org/3/tutorial/datastructures.html#more-on-lists).

Сортировка списка при помощи `.sort()`:
* по возрастанию (дефолтно)
* по убыванию (при помощи аргумента `reverse=True`)
* еще как-то (задав функцию – узнаем позже, как это делается)

````python
lst = [-1, 100, 34, 1, 45, -132]
lst.sort()  # меняет сам список!!
print(lst)
````

Подсчет числа вхождений элемента `x` в списке через `.count()`:

````python
comparative_linguistics = list('лингвистика-компаративистика')
comparative_linguistics.count('и')
````

Удаление первого элемента со значением `x` в списке через `.remove()`:

````python
lst.remove(-1)
print(lst)
````

````python
lst.remove(42) # если удалять то, чего и так нет, будет так
````

Удаление i-того элемента с его возвращением (дефолтно удаляет последний) через `.pop()`:

````python
lst.pop()
````

Удаление всех элементов списка через `.clear()`:

````python
lst.clear()
````

Вставка элемента на конкретную позицию в списке через `.insert()`:

````python
lst.insert(0, 'Новая поездка на море')
print(lst)
````

Добавление в конец элементов из переданного списка при помощи `.extend()`.

````python
lst.extend(['Каникулы', 'Новый год'])
print(lst)
````

### `.join()`

Если нам наоборот нужно из списка обратно собрать какую-то красивую строку для ответа, то используется метод строк `join()`. При этом список должен состоять из строк, чтобы все работало.

Синтаксис функции:

```
'{разделитель}'.join({итерируемый объект из строк})

````python
user_input = 'кофе кофе кофе'
items = user_input.split()
print('-'.join(items))
````

## Кортежи / `tuple`

**Кортеж – это упорядоченная неизменяемая коллекция объектов любых типов.** По сути – это просто неизменяемая версия списка. Он пригождается в более сложных структурах данных и, например, при кастомных функциях сортировок (об этом мы узнаем позднее).

### Как создать кортеж?

````python
# пустой кортеж
t1 = tuple()
t2 = ()
print(t1, t2)
````

````python
# готовый кортеж
t3 = tuple([1, 2, 3])
print(t3)
````

````python
# из одного элемента
t4 = ('Hello!', )
print(t4)
````

### Что можно делать с кортежем?

1. Индексироваться

````python
print(t3[0])
````

````python
# при попытке изменить элемент будет так
t3[0] = -1
````

2. Узнать его размер

````python
print(len(t4))
````

3. Посчитать число вхождений элемента `x` в кортеже

````python
t3.count(3)
````
