## Срезы строк (или списков)

Если нужно выделить из списка / строки не один элемент, а сразу несколько, то нужно использовать срезы.

Как это выглядит?

```python
# {строка}[{начальное значение}:{конечное значение}:{шаг}]

list[start:stop:step]
```

````python
list_of_chars = 'студент'

print(list_of_chars[1:]) # взять все символы, кроме первого
````

````python
letters = 'abcdefghijklmnopqrstuvwxyz'

n = 3

print(letters[2:15:(n-1)]) # взять символы с 2 по 15 (не включительно) с шагом в 2
````

В срезах можно значения пропускать.

**NB!** Одно значение в срезе до двоеточия – начало. Одно значение в срезе после двоеточия – конец. Два значения в срезе через двоеточие – это начало и конец.

````python
print(letters[10:]) # с 10 по конец
print(letters[:10]) # с начала до 10 (не включительно)
print(letters[5:10]) # с 5 по 10 (не включительно)
print(letters[-3:-1]) # с предпредпоследней до последней (не включительно)
print(letters[::2]) # с начала до конца с через 1
````

````python
letters = 'abcdefghijklmnopqrstuvwxyz'
letters[10:5:-1] # с 10 по 5 (не включительно) в обратном порядке
````

## Сравнение строк

Взгляните на код ниже. Подумайте, что он делает, и все ли, что он делает правильно.

````python
x = input('Введите первое целое число: ')
y = input('Введите второе целое число: ')

if x > y:
    print('Первое число больше второго!')
````

**NB!** Не забывайте переводить строки в числа, если оно нужно!

А почему вообще так происходит? Строки сравниваются по **лексикографическому порядку**, т.е. по алфавиту.

````python
print('11' > '1')
print('10' > '2')
print('2' > '1')
````

## Методы строк

### Разбить строку на части по символу / последовательности символов `split()`, а потом соединить при помощи `join()`

Опять же, если не уточнять, биться все будет по пробельным символам.

````python
s = 'Больше всего на свете я люблю спать'
s = s.split()

print(s)
print('_'.join(s))
````

### Поменять все буквы:
- на большие – `upper()`
- на маленькие – `lower()`
- на другой регистр – `swapcase()`

````python
print('абВ'.upper())
print('абВ'.lower())
print('абВ'.swapcase())
````

### Поменять часть букв на большие:
- `title()`
- `capitalize()`

````python
txt = 'сегодня мне на пары к 9:30(('
print(txt.title())
print(txt.capitalize())
print('')

txt = 'ПРИВЕТ, ХОчеШь ПоИГРаТЬ в ИГрУ?'
print(txt.title())
print(txt.capitalize())
print('')

txt = '101хак про питон...здесь'
print(txt.title())
print(txt.capitalize())
````

Подробнее – [тут](https://docs.python.org/3/library/stdtypes.html#string-methods).

````python
print('аб'.islower())
````

### Проверить, что строка:
- начинается с x? – `startswith()`
- заканчивается на x? – `endswith()`

````python
s = '1234532'

print(s.startswith('123'))
print(s.endswith('453'))
````

### Найти и / или заменить подстрочку в строке

- `find()` находит **первое** вхождение подстроки в строке и возвращает индекс первого совпадающего символа. Или -1, если совпадений нет. В качестве второго элемента можно передать позицию, с какого индекса можно начать поиск.

````python
coffee_s = 'пупупу, заварю-ка кофейку'

print(coffee_s.find('кофе'))
print(coffee_s.find('чай'))
````

````python
first_pu = coffee_s.find('пу')
second_pu = coffee_s.find('пу', first_pu + 1)
third_pu = coffee_s.find('пу', second_pu + 1)

print(first_pu, second_pu, third_pu)
````

- если вам не нужен номер элемента, а только да / нет, то это стоит делать при помощи `in`

````python
'кот' in 'бойкотировать'
````

- `replace(old, new)` находит все вхождения подстроки в строку и заменяет их на новую подстроку. В качестве третьего аргумента можно передать целое число – количество замен.

````python
'абракадабра'.replace('бра', 'люстра')
````

````python
# а тут что будет?

s = "aaaaa"
print(s.replace("aa", "b"))
````

````python
# Что тут может пойти не так?

print(s.count('а'))
````

### Убрать какие-то символы на краях при помощи `strip()`

Можно подать внутрь строку, содержащую все символы (подряд без запятых / пробелов), которые будут убираться на концах строки. Если оставить содержимое скобок пустым, то автоматически будут убираться все пробельные символы (пробелы, табы, переносы строк).

Есть две похожие функции:
- `rstrip()` удадяет только справа
- `lstrip()` удаляет только слева

````python
'11213'.strip('13') # что выведется?
````

````python
s = "    hello, honey!   "

print(s.__repr__()) # __repr__ тут нужен, чтобы увидеть скрытые символы
print(s.strip().__repr__()) # обрезать с двух сторон все пробельные символы
print(s.rstrip().__repr__()) # только справа
print(s.lstrip().__repr__()) # только слева
````
