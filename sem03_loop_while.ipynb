## **Цикл `while`**

**Циклы** — это конструкции, которые позволяют многократно выполнять один и тот же блок кода.

Цикл `while` работает, пока выполняется условие.

Он состоит из следующих частей:

1. **Ключевое слово `while`**  
2. **Условие** — логическое выражение, которое проверяется перед каждой итерацией цикла.
3. **Тело цикла** — блок кода с отступом, выполняемый, пока условие выполняется (истинно).

**Итерация** — это одно выполнение тела цикла во время выполнения программы.

```python
while Условие/я выполнения тела цикла:
    # четыре пробела
    Дейстие/я, которые нужно выполнить, ПОКА утверждение/я истинно/ы
```

Этот цикл удобен, если переменная в условии меняется в теле цикла. Пример, хочу возводить вводимые числа в квадрат, пока мне не введут 0:

````python
# пример пока не введут 0

n = int(input())

while n != 0:
    print("Принято")
    n = int(input())
````

Цикл может быть **бесконечным**. В таком случае прекратить его выполнение может оператор **break** (ну или ваша ОС).

````python
# пример с вводом пароля
true_password = 'abc'

while True:
    password = input()
    if true_password == password:
        print('Все в порядке! Заходите!')
        break
    else:
        print("Пароль неверный")
````

А что если у нас есть ограниченное число попыток ввода?

````python
attempts = 3
true_password = 'abc'

while attempts > 0:
    password = input()
    if true_password == password:
        print('Все в порядке! Заходите!')
        break
    else:
        attempts -= 1
        print("Пароль неверный")
        print(f"Оставшихся попыток: {attempts}")

if attempts == 0:
    print("Вы так и не смогли ввести верный пароль :(")
````
