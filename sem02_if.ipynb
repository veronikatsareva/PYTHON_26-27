## Условные конструкции

**Условные** конструкции позволяют нам проверять **условия** при помощи питона – верны или неверны какие-то выражения, важные для нашей задачи.

Общий синтаксис условной конструкции:

```
if условие:
    алгоритм, который выполняется
    только если верно условие (возвращает True)
elif условие:
    алгоритм
else:
    алгоритм, который выполняется всегда,
    когда не верны все предыдущие условия
```



*   в коде может быть сколько угодно условных конструкций (но эти условия проверяются независимо друг от друга / не знают друг про друга)
*   в условной конструкции может быть сколько угодно `elif`
*   условие в `elif` проверяется только тогда, когда неверно условие в предыдущих `if / elif`
*   внутри условия может быть любое выражение / любая функция, которая возвращает `True` или `False`
*   условные конструкции могут быть вложены друг в друга

````python
temp = 20

if temp > 25:
    print("Ну и жара!")
elif temp > 15:
    print("Теплая погода :)")
else:
    print("Ну и холодрыга...")
````

### Вложенные условия

Внутри условных операторов можно использовать любые блоки кода, например, все те же условные операторы, чтобы получить двойную вложенность.

Например, представьте, что вы пытаетесь проверить, является ли год високосным.



> **Как определить, что год високосный:**   
• Номер года делится на 4 без остатка, но при этом не делится на 100.  
• Номер года делится на 400 без остатка (например, 2000 год).

Ниже словами опишем структуру (да, не самой оптимальной) программы, которая определяет, является ли год високосным:

1. Если год делится на 4 без остатка, перейдите на шаг 2. В противном случае перейдите к выполнению действия 5.
2. Если год делится на 100 без остатка, перейдите на шаг 3. В противном случае перейдите к выполнению действия 4.
3. Если год делится на 400 без остатка, перейдите на шаг 4. В противном случае перейдите к выполнению действия 5.
4. Год високосный (366 дней).
5. Год не високосный (365 дней).

И, собственно, реализуем её:

````python
year = int(input())

if year % 4 == 0:  # можно еще просто if not year % 4:
    if year % 100 == 0:
        if year % 400 == 0:
            print('Год високосный')
        else:
            print('Год не високосный')
    else:
        print('Год високосный')
else:
    print('Год не високосный')
````

## Логические операторы: `and`, `or`, `not`

Логические операторы помогают формировать сложные условия внутри одной условной конструкции.

Оператор `and` – это **конъюнкция** / логическое «и» => хотим проверить, что несколько условий выполняются одновременно.

```
if условие1 and условие2:
    # => хотим, чтобы оба условия выполнялись одновременно
    алгоритм
```



````python
exam = 8
project = 7

if exam >= 8 and project >= 8:
    print("Обе оценки отличные!")
else:
    print("Какая-то из оценок (или обе) не отличные.")
````

Оператор `or` – это **дизъюнкция** / логическое «или» => хотим проверить, что хотя бы одно из условий выполняется.

```
if условие1 or условие2:
    # => хотим, чтобы хотя бы одно из условий выполнялось
    алгоритм
```



````python
exam = 8
project = 7

if exam >= 8 or project >= 8:
    print("Хотя бы одна оценка отличная!")
else:
    print("Ни одна из оценок не отличная.")
````

Оператор `not` – это оператор **отрицания** / логическое «не» => хотим проверить, что что-то не выполняется.

```
if not условие:
    алгоритм
```


````python
exam = 3

if not (exam >= 4):
    print("За экзамен неудовлетворительная оценка.")
else:
    print("За экзамен оценка выше или равна 4.")
````
